# 1. Introducción

**¿Qué es el proceso ETL?**
ETL son las siglas en inglés de Extract, Transform, Load (Extraer, Transformar y Cargar). Es un proceso fundamental en ingeniería de datos que permite mover datos desde múltiples fuentes, limpiarlos, formatearlos y cargarlos en un sistema de almacenamiento final para su posterior análisis o uso en modelos predictivos.

**Objetivo de la limpieza de datos**
El objetivo principal es identificar y corregir errores, inconsistencias, valores nulos o duplicados en el conjunto de datos. Esto asegura que la información sea confiable, precisa y útil para la toma de decisiones.

**Importancia de la calidad de los datos para análisis y Machine Learning**
La calidad de los datos es crucial. Como dice el adagio en ciencia de datos: "Garbage In, Garbage Out" (Basura entra, basura sale). Modelos de Machine Learning entrenados con datos de mala calidad producirán predicciones erróneas o sesgadas, sin importar cuán sofisticado sea el algoritmo. Un buen preprocesamiento garantiza resultados robustos y confiables.

## 3. Carga del dataset (Extract)

En esta etapa, extraemos los datos de nuestra fuente. Procedemos a cargar el dataset desde la carpeta `dataset/`. Es importante definir la ruta correctamente y utilizar pandas para leer el archivo CSV. Luego, mostraremos la cantidad de filas y columnas, y un vistazo a los primeros registros para entender su estructura básica.

```python
# Definimos la ruta al archivo del dataset.
# Asegúrate de que el archivo correcto se encuentre en esta ruta (puede variar según el nombre del archivo exacto)
file_path = Path("../dataset/Inventario_Departamental_Gases_Efecto_Invernadero_20260529.csv")

# Verificamos si el archivo existe antes de cargarlo
if file_path.exists():
    df = pd.read_csv(file_path)
    print(f"El dataset tiene {df.shape[0]} filas y {df.shape[1]} columnas.\n")
    print("Primeros registros del dataset:")
    display(df.head())
else:
    print(f"El archivo no fue encontrado en la ruta: {file_path}")
    # Dataset de ejemplo para demostrar el notebook si no se encuentra el archivo
    print("Creando un dataset de ejemplo para la demostración...")
    data = {
        ' Nombre Departamento ' : ['Bogotá', 'Antioquia', 'Valle', 'Bogotá', 'Meta'],
        'CAP': [1, 2, 3, 1, 4],
        'CAT': ['A', 'B', 'C', 'A', 'D'],
        'SUBNROM': [10, 20, 30, 10, 40],
        'NUM': [100, 200, 300, 100, 400],
        'REPORTAR': ['SI', 'NO', 'SI', 'SI', 'NO'],
        'NIVEL': [1, 2, 1, 1, 3],
        'RUBRO': ['X', 'Y', 'Z', 'X', 'W'],
        'SECTORES': ['S1', 'S2', 'S3', 'S1', 'S4'],
        'Emisiones': [150.5, np.nan, 80.2, 150.5, np.nan]
    }
    df = pd.DataFrame(data)
    print(f"El dataset de ejemplo tiene {df.shape[0]} filas y {df.shape[1]} columnas.\n")
    print("Primeros registros del dataset:")
    display(df.head())
```

## 5. Transformación de los datos (Transform)

### Eliminación de columnas innecesarias

En el análisis de datos, mantener información irrelevante consume recursos y puede generar ruido en los modelos. Las columnas `CAP`, `CAT`, `SUBNROM`, `NUM`, `REPORTAR`, `NIVEL`, `RUBRO` y `SECTORES` son identificadas como variables administrativas o de clasificación interna que no aportan valor predictivo ni analítico para el objetivo general de evaluar las emisiones.

Por lo tanto, procederemos a eliminarlas.

```python
# Lista de columnas que deseamos eliminar, exactamente como se solicita
columnas_a_eliminar = ['CAP', 'CAT', 'SUBNROM', 'NUM', 'Reportar', 'Nivel', 'Rubro', 'Sector (Gráfica sectorial)', 'Sector (Colores histograma)']

# Verificamos cuáles de estas columnas realmente existen en el dataset
columnas_existentes = [col for col in columnas_a_eliminar if col in df.columns]
columnas_no_encontradas = [col for col in columnas_a_eliminar if col not in df.columns]

print("--- Verificación de columnas a eliminar ---")
print(f"Columnas encontradas y que serán eliminadas: {columnas_existentes}")
if columnas_no_encontradas:
    print(f"Columnas NO encontradas en el dataset: {columnas_no_encontradas}")

# Eliminamos las columnas encontradas
df = df.drop(columns=columnas_existentes)
print("\nLas columnas han sido eliminadas.")
```

### Transformación de las variables MOD y Categorías

Durante la exploración inicial del conjunto de datos se identificó que la columna **categorias** contiene un nivel de detalle muy específico para cada fuente de emisión de gases de efecto invernadero.

Ejemplos:

- 1A5bi - Móvil (componente aviación)
- 1C1a - Gasoductos
- 1C1b - Embarcaciones
- 2A - Industria de los minerales
- 3A1 - Fermentación entérica

Aunque este nivel de detalle puede ser útil para análisis especializados, genera una gran cantidad de categorías y dificulta el análisis agregado por sector económico o ambiental.

---

##### Uso de la variable MOD

La variable **MOD** representa la clasificación principal de los sectores de emisión según la metodología del IPCC (Panel Intergubernamental sobre Cambio Climático).

La correspondencia es la siguiente:

| MOD | Sector Principal |
|------|----------------|
| 1 | Energía |
| 2 | Procesos Industriales |
| 3 | Agricultura |
| 4 | Silvicultura y Otros Usos del Suelo |
| 5 | Residuos |

---

#### Estrategia aplicada

Se creó una nueva variable denominada **sector_principal**, utilizando la información contenida en la columna **MOD**.

De esta manera, las categorías específicas quedan agrupadas dentro de uno de los cinco grandes sectores de emisión.

Por ejemplo:

| MOD | Categoría Original | Sector Principal |
|------|-------------------|------------------|
| 1 | 1A5bi - Móvil (componente aviación) | Energía |
| 1 | 1C1a - Gasoductos | Energía |
| 2 | 2A - Industria de los minerales | Procesos Industriales |
| 3 | 3A1 - Fermentación entérica | Agricultura |

---

#### Justificación

La creación de la variable **sector_principal** permite:

- Reducir la complejidad del conjunto de datos.
- Facilitar análisis exploratorios y visualizaciones.
- Realizar comparaciones entre grandes sectores económicos.
- Simplificar futuros modelos de Machine Learning.
- Mantener la información detallada de las categorías originales para análisis posteriores.

Por esta razón, se decidió conservar ambas variables:

- **sector_principal** → nivel agregado.
- **categorias** → nivel detallado.

Esta estrategia permite realizar análisis tanto generales como específicos sin perder información relevante.

```python
# Diccionario de equivalencias
mapa_mod = {
    1: "Energía",
    2: "Procesos Industriales",
    3: "Agricultura",
    4: "Silvicultura y Otros Usos del Suelo",
    5: "Residuos"
}

df["sector_principal"] = df["MOD"].map(mapa_mod)

# Renombrar la categoría específica
df.rename(
    columns={"categorias": "subcategoria"},
    inplace=True
)
```

### Validación posterior

Verificamos que los cambios se hayan aplicado correctamente, revisando la nueva lista de columnas y las dimensiones del dataset.

```python
print("--- Validación posterior a la eliminación ---")
print("Nuevas dimensiones del dataset:")
print(f"Filas: {df.shape[0]}, Columnas: {df.shape[1]}\n")
print("Columnas restantes:")
print(df.columns.tolist())
```

---
## 6. Limpieza básica adicional

### Valores duplicados

Los registros duplicados pueden sesgar las métricas y los modelos al contar varias veces una misma observación. Vamos a detectar cuántos existen y eliminarlos de inmediato si los hay.

```python
# Detectar duplicados
cantidad_duplicados = df.duplicated().sum()
print(f"Se encontraron {cantidad_duplicados} registros duplicados.")

if cantidad_duplicados > 0:
    # Eliminar duplicados
    df = df.drop_duplicates()
    print("Los registros duplicados han sido eliminados.")
else:
    print("No se requirió eliminación.")
```

### Valores nulos

Identificamos la presencia de valores nulos (faltantes). En este momento **solo generaremos el diagnóstico y no realizaremos una imputación automática**. La imputación correcta se debe tomar basada en un análisis posterior con expertos de negocio.

**Posibles estrategias de tratamiento para el futuro:**
*   **Eliminación:** Si los nulos representan un porcentaje ínfimo (ej. < 5%), borrar la fila puede ser razonable.
*   **Imputación estadística:** Rellenar con la media, mediana o moda.
*   **Imputación por Machine Learning:** Predecir el valor usando variables relacionadas.
*   **Marcado explícito:** Indicar en una nueva columna que el valor fue nulo (ej. si los datos faltantes no son aleatorios).

```python
print("--- Diagnóstico de valores nulos ---")

nulos_por_columna = df.isnull().sum()

print("Cantidad de valores nulos por columna:")
display(nulos_por_columna)

# Cantidad de registros antes de la limpieza
filas_antes = len(df)

# Eliminar registros donde MOD sea nulo
df = df.dropna(subset=['MOD'])

# Cantidad de registros eliminados
filas_despues = len(df)
eliminados = filas_antes - filas_despues

print(f"\nRegistros eliminados por tener MOD nulo: {eliminados}")
print(f"Registros restantes: {filas_despues}")
```

### Estandarización de nombres de columnas

Tener nombres de columnas estandarizados facilita enormemente la programación y previene errores técnicos. Aplicaremos el siguiente formato a los encabezados:
1.  Convertir todas las letras a minúsculas.
2.  Quitar espacios en blanco al inicio o al final del nombre.
3.  Reemplazar los espacios intermedios por guiones bajos (`_`).

```python
print("--- Estandarización de columnas ---")
print("Columnas originales:", df.columns.tolist())

# Aplicamos strip, lower y replace utilizando list comprehension
df.columns = [col.strip().lower().replace(' ', '_') for col in df.columns]

print("\nColumnas estandarizadas:", df.columns.tolist())
```

Antes de guardar el dataset y finalizar la etapa ETL, damos un último vistazo para asegurar que todas las transformaciones se aplicaron adecuadamente.

```python
print("--- Estado final del Dataset ---")
print(f"Dimensiones finales: {df.shape[0]} filas, {df.shape[1]} columnas.\n")
print(f"Número total de registros: {df.shape[0]}")
print(f"Número total de columnas: {df.shape[1]}\n")

print("--- Tipos de datos finales ---")
print(df.dtypes)

print("\n--- Vista previa del dataset limpio ---")
display(df.head())
```

## 8. Carga y almacenamiento (Load)

El paso final del proceso ETL es el Load (Carga). Guardaremos nuestro DataFrame limpio y transformado en un archivo CSV dentro de la carpeta designada para datos procesados. Esto asegura que no sobrescribamos los datos originales (Raw Data) y dejamos la información lista para consumo de modelos de Machine Learning y dashboards analíticos.

Se creará automáticamente la carpeta `dataset/processed/` en caso de que no exista previamente.

```python
# Definimos la ruta de salida
output_dir = Path("../dataset/processed")
output_file = output_dir / "dataset_limpio.csv"
output_file_excel = output_dir / "dataset_limpio.xlsx"

# Creamos el directorio si no existe (crea también carpetas padre si es necesario)
output_dir.mkdir(parents=True, exist_ok=True)

# Guardamos el dataframe sin el índice para evitar una columna extra indeseada
df_excel = df.copy()

df.to_csv(output_file, index=False)
df_excel.to_excel(output_file_excel, index=False)

print(f"El dataset limpio ha sido guardado exitosamente en: {output_file}")
```
